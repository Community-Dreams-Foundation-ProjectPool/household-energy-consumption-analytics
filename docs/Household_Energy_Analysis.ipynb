```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import streamlit as st

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Plotly imported successfully")
print("Streamlit:", st.__version__)
print("Environment setup successful")
```

```python
df = pd.read_csv(
    "household_power_consumption.txt",
    sep=";",
    na_values="?"
)

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# combined datetime field
df["Datetime"] = pd.to_datetime(
    df["Date"] + " " + df["Time"],
    format="%d/%m/%Y %H:%M:%S"
)

df.head()
```

```python
#Create the time-based columns you will eventually use as dashboard filters

df["Year"] = df["Datetime"].dt.year
df["Month"] = df["Datetime"].dt.month
df["Day"] = df["Datetime"].dt.day
df["Hour"] = df["Datetime"].dt.hour
df["DayOfWeek"] = df["Datetime"].dt.day_name()
```

```python
#For Season

def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Fall"

df["Season"] = df["Month"].apply(get_season)
```

Now the dataset is ready for filtering by hour, day, month, and season.

```python
# Check Missing values

df.isnull().sum()
```

```python
# Save Clean Version of the Data

df.to_csv("household_energy_cleaned.csv", index=False)
```
